- 如何在竞品检测中使用RFM模型
  - R-Recency   最近一次进入top榜单
  - F-Frequency   进入top榜单次数
  - M-top10    高排名榜单次数/近3天高排名榜单次数

- 针对距离观察日三个月内，将竞品状态分为以下几种类型：重点关注、重点调研、重点保持、一般调研、一般关注、暂不关注；将竞品分为老品，新品两类

| 竞品分类         | 最近一次进入top榜单 | 进入top榜单次数 | **高排名榜单次数** | 近3天高排名榜单次数 | **策略建议**                     |
| ---------------- | ------------------- | --------------- | ------------------ | ------------------- | -------------------------------- |
| **持续关注竞品** | **高**              | **高**          | **高**             | **高**              | **心中有数，最重要竞品**         |
| **重点调研老品** | **高**              | **高**          | **低**             | **高**              | **老品要发力，需运营关注**       |
| **重点调研新品** | **高**              | **低**          | **高**             | **高**              | **需要运营重点研究**             |
| **重点保持老品** | **高**              | **高**          | **高**             | **低**              | **近期排名下降**                 |
| 一般调研新品     | 高/低               | 低              | 低                 | 高                  | 新品期产品，数据监控             |
| 一般调研老品     | 低                  | 高              | 高                 | 低                  | 过了新品期后运营乏力，可数据监控 |
| 一般关注老品     | 高/低               | 高              | 低                 | 低                  | 竞争力相对较弱产品               |
| 暂不关注产品     | 其它情况            |                 |                    |                     |                                  |

- 我们的数据会每天采集多次，所以每天每件商品上榜次数可能不唯一，这里取排名最高的一次

```python
# 
df_competitive_rfm =df_competitive[['id','rank','date_d']].groupby(['id','date_d'],as_index=False).agg({'rank':'min'})
df_competitive_rfm.columns=['id','date_d','rank_min']
```

```python
# date_d转换为日期时间类型
df_competitive_rfm['date_d'] = pd.to_datetime(df_competitive_rfm['date_d'])
```

```python
df_competitive_rfm.info()
```

- 判断是否进入Top10，进入Top10作为价值高低的判定依据

```python
# 排行进入前10名记为1 否则是0

df_competitive_rfm['isTop10'] =df_competitive_rfm['rank_min'].apply(lambda x:1 if x<=10 else 0)
```

- 筛选最近90天的数据来进行RFM打分，数据集中最后的日期是2020年3月8日，我们以这一天为基准

```python
calc_date = df_competitive_rfm['date_d'].max() # 数据集中最后一天作为计算的日期
df_competitive_rfm['date_diff'] =(calc_date - df_competitive_rfm['date_d']).dt.days # 每一天跟计算这一天 算天数差值
df_competitive_rfm_90=df_competitive_rfm[df_competitive_rfm['date_diff']<=90] # 筛选最近90天的数据
df_competitive_rfm_90
```

```python
df_competitive_rfm_90.shape
df_competitive_rfm_90.describe()
```

### 3、计算RFM

- RFM指标计算方法
  - R-Recency 最后一次出现在热销榜距离当前时间的天数
  - F-Frequecy 上榜频次
  - M1-上榜top10的次数
  - M2-近3天进入top10次数
- 计算Recency 最后一次top排名距离现在的时间间隔

```python
r_data = df_competitive_rfm_90.groupby(['id'])[['date_diff']].min()
r_data.columns=['R']
r_data
```

- 计算F的原始数据（聚合之后的结果），Frequency 有多少次出现在热销榜上

```python
f_data = df_competitive_rfm_90.groupby(['id'])[['rank_min']].count()
f_data.columns=['F']
f_data
```

```python
df_competitive_rfm_90.columns
```

- M1 上TOP10榜单次数，进入top10 打1，没有进入标记为0, 统计1的次数

```python
m1_data = df_competitive_rfm_90.groupby(['id'])[['isTop10']].sum()
m1_data.columns=['M1']
m1_data
```

- M2 近三天上top10次数，体现近期

```python
temp_df = df_competitive_rfm_90[df_competitive_rfm_90['date_diff']<=2]
m2_data = temp_df.groupby(['id'])[['isTop10']].sum()
m2_data.columns=['M2']
```

- 将RFM的原始数据合并起来

```python
m2_data
```

```python
rfm =pd.concat([r_data,f_data,m1_data,m2_data],axis=1)
rfm.fillna(0,inplace=True)
rfm.describe()
```

### 4、 RFM打分

| 按照运营需求打分 | 最近一次上榜-R | 上榜次数-F | 进入top10榜单次数-M1 | 近3天进入top10次数-M2 |
| ---------------- | -------------- | ---------- | -------------------- | --------------------- |
| 1                | 80天以上       | 小于20次   | 小于5次              | 1次                   |
| 2                | 30-80天        | 20-40次    | 5-20次               | 2次                   |
| 3                | 15-30天        | 40-60次    | 20-60次              | 3次                   |
| 4                | 3-15天         | 60-80次    | 60-80次              |                       |
| 5                | 3天以内        | 80次以上   | 80次以上             |                       |

- 通过pd.cut方法，将用户分层并打上标签， 打分规则见上面表格

```python
rfm.shape[0]
```

```python
r_bins = [0,3,15,30,80,999]
labels = [i for i in range(len(r_bins)-1,0,-1)]
rfm['R_score'] = pd.cut(rfm['R'],bins=r_bins,labels=labels,include_lowest=True).astype(int)
f_bins = [0,20,40,60,80,999]
labels = [i for i in range(1,len(f_bins))]
rfm['F_score'] = pd.cut(rfm['F'],bins=f_bins,labels=labels,include_lowest=True).astype(int)
m1_bins = [0,5,20,60,80,999]
labels = [i for i in range(1,len(m1_bins))]
rfm['M1_score'] = pd.cut(rfm['M1'],bins=m1_bins,labels=labels,include_lowest=True).astype(int)
m2_bins = [0,1,2,3,100]
labels = [i for i in range(1,len(m2_bins))]
rfm['M2_score'] = pd.cut(rfm['M2'],bins=m2_bins,labels=labels,include_lowest=True).astype(int)
```

```python
rfm['rfm_labels']
```

```python
[i for i in range(1,len(f_bins))]
```

- 然后我们对不同指标1~5分计算平均值，高于/低于平均值的，进一步二分 每个维度均标记为高或低

```python
rfm['R_label'] = rfm['R_score'].apply(lambda x:'高' if x>rfm['R_score'].mean() else '低')
rfm['F_label'] = rfm['F_score'].apply(lambda x:'高' if x>rfm['F_score'].mean() else '低')
rfm['M1_label'] = rfm['M1_score'].apply(lambda x:'高' if x>rfm['M1_score'].mean() else '低')
rfm['M2_label'] = rfm['M2_score'].apply(lambda x:'高' if x>rfm['M2_score'].mean() else '低')
```

```python
rfm['rfm_label'] = rfm['R_label']+rfm['F_label']+rfm['M1_label']+rfm['M2_label']
```

- 将不同的类别，打上不同的标签，不同的标签，就是不同的产品类型，也就是不同的对待策略

```python
# 分层规则
rfm2cat={
    '高高高高':'持续关注竞品',
    '高高低高':'重点调研老品',
    '高低高高':'重点调研新品',
    '高高高低':'重点保持老品',
    '低低低高':'一般调研新品',
    '高低低高':'一般调研新品',
    '低高高低':'一半调研老品',
    '高高低低':'一般关注老品',
    '低高低低':'一般关注老品',
}
```

```python
# 按照标签分类，查看每类商品的数量， 如果有没打上标签的， 用暂不关注产品 填充
```

```python
rfm['product_label'] = rfm['rfm_label'].map(rfm2cat)
```

```python
rfm['product_label'].fillna('暂不关注产品',inplace=True)
```
