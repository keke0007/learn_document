```python
import pandas as pd
import statsmodels.stats.api as sms
# 计算参与实验的人数
effect_size = sms.proportion_effectsize(0.13,0.15) # 0.13 是当前的转化率, 0.15 目标转化率
required_n = sms.NormalIndPower().solve_power(effect_size,power=0.8,alpha=0.05,ratio=1) #power 1-β 0.8   alpha 显著性水平, 一般就是5%
```

```python
from math import ceil
# ceil向上去整数
required_n =ceil(required_n)
required_n
```

```python
users = session_counts[session_counts<2].index
ab_data=ab_data[ab_data['user_id'].isin(users)]
```

```python
# 从数据中采样 4720个
control_sample = ab_data[ab_data['group']=='control'].sample(n=required_n,random_state=13) # 对照组 看老页面
test_sample = ab_data[ab_data['group']=='treatment'].sample(n=required_n,random_state=13) # 实验组 看新页面
control_sample['converted'].mean()
```
