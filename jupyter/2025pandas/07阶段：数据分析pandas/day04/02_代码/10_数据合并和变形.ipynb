```python
import pandas as pd
df1 = pd.DataFrame([[1, 2, 3], [1, 10, 20], [5, 6,7], [3, 9, 0], [8, 0, 3]], columns=['x1', 'x2', 'x3'])
df2 = pd.DataFrame([[1, 2], [1, 10], [1, 3], [4, 6], [3, 9]], columns=['x1', 'x4'])
print(df1)
print(df2)
```

## df.append() 了解

```python
df1.append(df2) # df1后面追加一些新的行, df1和df2的数据, 如果列名相同就能够追加到一起, 如果列名不同, 拼接不上的部分用NaN填充
```

```python
df1.append(df2, ignore_index=True) # ignore_index 拼接之后, 忽略索引 索引会从0开始一路数下去  默认会保留之前的索引
```

## pd.concat  需要掌握

```python
pd.concat([df1, df2], ignore_index=True)
```

```python
pd.concat([df1, df2], axis=1)
```

```python
pd.concat([df1.set_index(['x1']), df2], axis=1)
```

```python
df3 = pd.DataFrame([[1, 2], [1, 10], [1, 3], [4, 6], [3, 9]], columns=['x1', 'x4'],index=['a', 'b', 'c', 'd', 'e'])
```

```python
pd.concat([df1, df3], axis=0, join='outer')
```

## merge连接 和SQL的join基本相同, 需要掌握

```python
df1
```

```python
df2
```

```python
df1.merge(df2,on='x1',how='left')
```

```python
df1.merge(df2,on='x1',how='right')
```

```python
df1.merge(df2,on='x1',how='inner')
```

```python
df1.merge(df2,on='x1',how='outer')
```

```python
df1.merge(df2,left_on='x3',right_on='x1',how='left',suffixes=['_left','_right'])
# 如果要关联的两张表使用的列名不一样, 通过left_on 和 right_on 来指定
#  left_on 左表出哪一列用于关联
#  right_on  右表出哪一列用于关联
#  suffixes=['_left','_right']  两张表关联之后, 如果有相同的列名, 通过这个参数可以指定后缀 默认是['_x','_y']
```

```python
## join拼接 
# concat 和 merge  的功能稍微调整就可以实现join的效果
```

```python
df1.join(df3,rsuffix='_right',lsuffix='_left',how='outer')
```

```python
pd.concat([df1,df2],axis=1)
```

```python
df1.join(df2,on='x1' ,rsuffix='_right',lsuffix='_left')
```
