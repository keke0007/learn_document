## 1 不同产品的销售情况

### 1.1 统计不同产品的销售总量


```python
uniqlo.groupby(['产品名称'])[['订单数量','销售金额']].sum()
```

### 1.2 按城市统计不同产品类别的销售总量

```python
uniqlo.pivot_table(index='城市',columns='产品名称',values=['订单数量','销售金额'],aggfunc='sum')
```

### 1.3 进一步按城市以及线上线下分析

```python
uniqlo.pivot_table(index=['城市','销售渠道'],columns='产品名称',values=['订单数量','销售金额'],aggfunc='sum')
```

## 2 用户习惯使用哪种方式进行消费 (线上还是线下)

```python
online_list = ['上海','广州','武汉','西安','重庆']
uniqlo_online = uniqlo[uniqlo['城市'].isin(online_list)]
uniqlo_online.pivot_table(index=['销售渠道','性别'],columns='产品名称',values=['订单数量'],aggfunc='sum')
```

```python
uniqlo_online.pivot_table(index='销售渠道',columns='性别',values='订单数量',aggfunc='sum',margins=True)
```

```python
uniqlo_online.pivot_table(index='销售渠道',columns='年龄段',values='订单数量',aggfunc='sum',margins=True)
```

- 有些城市没有线上, 只有线下  (北京 南京 杭州 深圳)

## 3 用户消费时间习惯 (周间还是周末)

```python
temp_df = uniqlo.groupby(['消费时间'])[['订单数量']].sum()
```

```python
temp_df['days'] = [5,2]
```

```python
temp_df['订单数量']/temp_df['days']
```

```python
temp_df = uniqlo.pivot_table(index='城市',columns='消费时间',values='订单数量',aggfunc='sum')
temp_df['周间平均'] = temp_df['Weekday']/5
```

```python
temp_df['周末平均'] = temp_df['Weekend']/2
```

```python
temp_df['周间平均']/temp_df['周末平均']
```
