## apply 自定义函数

```python
temp_df = df.head().copy()
temp_df
```

```python
def func(area_str,arg1,arg2):
    print('调用了func方法====',area_str)
    if area_str =='广安门租房':
        return arg1
    elif area_str =='天通苑租房':
        return arg2
    else:
        return area_str
```

```python
result_list = []
for area_str in temp_df['区域'].to_list():
    result_list.append(func(area_str))
```

```python
pd.Series(result_list,name='区域')
```

```python
temp_df['区域'].apply(func,args=('西城租房','昌平租房'))
```

```python
def func1(s):
    if s['区域'] == '望京租房':
        s['户型']='1室1厅'
    elif s['区域'] == '天通苑租房':
        s['户型']='3室1厅'
    return s
```

```python
temp_df2.apply(func1,axis=1)
```

```python
# 创建一个空白的DataFrame用于保存结果
result_df = pd.DataFrame()
# 遍历行索引
for ind in temp_df2.index:
    # 获取了每一行数据
    s = temp_df2.loc[ind]
    # 通过自定义函数, 对每一行数据进行处理
    s = func1(s)
    # 处理之后的series 添加到dataFrame
    result_df =result_df.append(s)
```

```python
result_df
```

```python
def func2(s):
    # print(s)
    # print('=======',s.name)
    if s.name=='地址':
        return s+'小区'
    elif s.name=='面积':
        return s/2
    else:
        return s
        
```

```python
temp_df2.apply(func2)
```

```python
temp_df2.apply(func1,axis=1)
```

```python
def func1(s,arg1,arg2):
    if s['区域'] == '望京租房':
        s['户型']=arg1
    elif s['区域'] == '天通苑租房':
        s['户型']=arg2
    return s
temp_df2.apply(func1,axis=1,args=('1室0厅','3室1厅'))
```

```python
def func3(value):
    print(value)
    print('=======apply map =========')
    if value=='天通苑租房':
        return '昌平租房'
    elif value ==8000:
        return value+500
    else:
        return value
temp_df2.applymap(func3)
```
