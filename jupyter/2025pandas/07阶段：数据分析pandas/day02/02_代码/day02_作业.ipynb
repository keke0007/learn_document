```python
import pandas as pd
movie_df = pd.read_csv('C:/Develop/深圳42/data/movie.csv')
```

- 1 把数据中所有 的James Cameron 替换成中文 詹姆斯·卡梅隆 

```python
movie_df.replace('James Cameron', '詹姆斯·卡梅隆', inplace=True)
```

```python
movie_df['director_name'].value_counts()
```

- 2 查找数据集中上映年份(title_year)最近一年,   内容分级(content_rating)为 Not Rated 和  Unrated , imdb评分大于 6.6分的电影

```python
movie_df[(movie_df['title_year'] == movie_df['title_year'].max()) & (movie_df['content_rating'].isin(['Not Rated','Unrated']) & (movie_df['imdb_score']>6.6))]
```

- 3 从数据集中选出如下列:
    - 电影名称 (movie_title), 导演姓名 (director_name), 主演名字(actor_1_name), 上映年份(title_year), imdb_score
    - 找到数据集中最近3年 评分最高的电影名称, 导演姓名, 主演名字, imdb_score分数

```python
movie_df_2 = movie_df[['movie_title', 'director_name', 'actor_1_name', 'title_year','imdb_score']]
latest_3year = movie_df_2['title_year'].value_counts().sort_index(ascending=False).head(3).index
movie_df_2[movie_df_2['title_year'].isin(latest_3year)].sort_values(by='imdb_score', ascending=False).head(1)
```

- 4 找到2016年的imdb_score分数最高的5部电影, 给这5部电影构成的数据集添加一列, 取值为'年度最佳电影''

```python
head_5df = movie_df_2[movie_df_2['title_year']==2016].sort_values(by='imdb_score', ascending=False).head(5)
head_5df['电影类型'] ='年度最佳电影'
```

```python
head_5df
```

- 5 给电影数据集添加一列, 利润 , 用gross 收入 - budget 预算  放到 budget列后面

```python
movie_df['profite']=movie_df['gross']-movie_df['budget']
movie_df.insert(loc=23,column='profite',value=movie_df['gross']-movie_df['budget'],allow_duplicates=True)
```

- 6 把数据中的movie_facebook_likes 列删除

```python
movie_df.drop('movie_facebook_likes', axis=1, inplace=True)
```

```python
movie_df['movie_facebook_likes']
```

- 7 筛选出2016年的电影作为一个新的df, 把2016年的电影中, imdb评分数低于7分的电影删掉

```python
drop_list = movie_df_2[(movie_df_2['title_year']==2016) & (movie_df_2['imdb_score']<7)].index
movie_df_2.drop(drop_list)
```
