```python
def func(s):
    print(s)
    print(type(s))    
    print('=====================')
    # 每一组选出价格>3000的 对这部分求平均
    result = s[s<12000].mean()
    return result
## 自定义聚合函数
df.groupby(['区域'])['价格'].agg(func)
```

```python
# 价格低于4500 低    4500~8500 中  >8500 高
```

```python
def get_price_label(price):
    if price<=4500:
        return '低'
    elif price<=8500:
        return '中'
    else:
        return '高'
    
df['价格'].apply(get_price_label).value_counts()
```
