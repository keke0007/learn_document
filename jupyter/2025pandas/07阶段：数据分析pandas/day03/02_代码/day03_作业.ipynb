- ⑤ 乘客姓名中 有些人的名字里包含了特殊的信息 ['Lady', 'Countess','Capt', 'Col','Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 添加一列 身份 'Rare' / 'Norm'

```python
titanic_df['Name'].head(10)
```

```python
list_1 = ['Lady.', 'Countess.','Capt.', 'Col.','Don.', 'Dr.', 'Major.', 'Rev.', 'Sir.', 'Jonkheer.', 'Dona.']
def func(name):
    # 遍历 特殊称呼列表
    for str in list_1:
        # 如果 在名字中包含了一个特殊称呼
        if str in name:
            # 返回 Rare
            return 'Rare'
        # else:
        #     return ''
    # 所有的特殊称呼都遍历一圈, 没有在名字里, 返回Norm
    return 'Norm'
titanic_df['Name'].apply(func).value_counts()
```

```python
titanic_df['cate'] = titanic_df['Name'].apply(func)
```

```python
titanic_df[titanic_df['cate']=='Rare']
```

- ⑥ 将年龄分成几组 0~18 未成年 18~35 青年 35~60 中年 >60 老年 并计算不同年龄段的生还率

```python
titanic_df['年龄段'] =pd.cut(titanic_df['Age'], bins=[0,18,35,60,100], labels=['未成年', '青年','中年','老年'])
```

```python
titanic_df['年龄段'].value_counts()
```

```python
titanic_df.groupby('年龄段')['Survived'].mean()
```
